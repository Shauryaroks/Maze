# Comparison: Static vs Dynamic Hyperparameters

The following cells compare the results of the static (original) and dynamic (real-time tuned) hyperparameter runs for ACO, GA, and Hybrid metaheuristics on both A* and Dijkstra routers. We use the two CSVs: `full_benchmark.csv` (static) and `full_benchmark_dynamicheuristics.csv` (dynamic).


```python
# Load both static and dynamic hyperparameter benchmark results
df_static = pd.read_csv("full_benchmark.csv")
df_dynamic = pd.read_csv("full_benchmark_dynamicheuristics.csv")
df_static['type'] = 'Static'
df_dynamic['type'] = 'Dynamic'
```

## A* Router: Path Lengths and Execution Times

The following plots compare the path lengths and execution times for ACO, GA, and Hybrid metaheuristics using static and dynamic hyperparameters on the A* router.

```python
plt.figure(figsize=(12,6))
plt.plot(df_static['run'], df_static['aco_astar_length'], label='ACO (A*) Static', color='blue', linestyle='--')
plt.plot(df_dynamic['run'], df_dynamic['aco_astar_length'], label='ACO (A*) Dynamic', color='blue')
plt.plot(df_static['run'], df_static['ga_astar_length'], label='GA (A*) Static', color='red', linestyle='--')
plt.plot(df_dynamic['run'], df_dynamic['ga_astar_length'], label='GA (A*) Dynamic', color='red')
plt.plot(df_static['run'], df_static['hybrid_astar_length'], label='Hybrid (A*) Static', color='magenta', linestyle='--')
plt.plot(df_dynamic['run'], df_dynamic['hybrid_astar_length'], label='Hybrid (A*) Dynamic', color='magenta')
plt.xlabel('Run')
plt.ylabel('Path Length (meters)')
plt.title('A* Router: Path Lengths (Static vs Dynamic Hyperparameters)')
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(12,6))
plt.plot(df_static['run'], df_static['aco_astar_time'], label='ACO (A*) Static', color='blue', linestyle='--')
plt.plot(df_dynamic['run'], df_dynamic['aco_astar_time'], label='ACO (A*) Dynamic', color='blue')
plt.plot(df_static['run'], df_static['ga_astar_time'], label='GA (A*) Static', color='red', linestyle='--')
plt.plot(df_dynamic['run'], df_dynamic['ga_astar_time'], label='GA (A*) Dynamic', color='red')
plt.plot(df_static['run'], df_static['hybrid_astar_time'], label='Hybrid (A*) Static', color='magenta', linestyle='--')
plt.plot(df_dynamic['run'], df_dynamic['hybrid_astar_time'], label='Hybrid (A*) Dynamic', color='magenta')
plt.xlabel('Run')
plt.ylabel('Execution Time (seconds)')
plt.title('A* Router: Execution Times (Static vs Dynamic Hyperparameters)')
plt.legend()
plt.show()
```

## Dijkstra Router: Path Lengths and Execution Times

The following plots compare the path lengths and execution times for ACO, GA, and Hybrid metaheuristics using static and dynamic hyperparameters on the Dijkstra router.

```python
plt.figure(figsize=(12,6))
plt.plot(df_static['run'], df_static['aco_dijkstra_length'], label='ACO (Dijkstra) Static', color='blue', linestyle='--')
plt.plot(df_dynamic['run'], df_dynamic['aco_dijkstra_length'], label='ACO (Dijkstra) Dynamic', color='blue')
plt.plot(df_static['run'], df_static['ga_dijkstra_length'], label='GA (Dijkstra) Static', color='red', linestyle='--')
plt.plot(df_dynamic['run'], df_dynamic['ga_dijkstra_length'], label='GA (Dijkstra) Dynamic', color='red')
plt.plot(df_static['run'], df_static['hybrid_dijkstra_length'], label='Hybrid (Dijkstra) Static', color='magenta', linestyle='--')
plt.plot(df_dynamic['run'], df_dynamic['hybrid_dijkstra_length'], label='Hybrid (Dijkstra) Dynamic', color='magenta')
plt.xlabel('Run')
plt.ylabel('Path Length (meters)')
plt.title('Dijkstra Router: Path Lengths (Static vs Dynamic Hyperparameters)')
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(12,6))
plt.plot(df_static['run'], df_static['aco_dijkstra_time'], label='ACO (Dijkstra) Static', color='blue', linestyle='--')
plt.plot(df_dynamic['run'], df_dynamic['aco_dijkstra_time'], label='ACO (Dijkstra) Dynamic', color='blue')
plt.plot(df_static['run'], df_static['ga_dijkstra_time'], label='GA (Dijkstra) Static', color='red', linestyle='--')
plt.plot(df_dynamic['run'], df_dynamic['ga_dijkstra_time'], label='GA (Dijkstra) Dynamic', color='red')
plt.plot(df_static['run'], df_static['hybrid_dijkstra_time'], label='Hybrid (Dijkstra) Static', color='magenta', linestyle='--')
plt.plot(df_dynamic['run'], df_dynamic['hybrid_dijkstra_time'], label='Hybrid (Dijkstra) Dynamic', color='magenta')
plt.xlabel('Run')
plt.ylabel('Execution Time (seconds)')
plt.title('Dijkstra Router: Execution Times (Static vs Dynamic Hyperparameters)')
plt.legend()
plt.show()
```

```python
# Combine static and dynamic benchmark CSVs into one file
import pandas as pd

df_static = pd.read_csv("full_benchmark.csv")
df_dynamic = pd.read_csv("full_benchmark_dynamicheuristics.csv")
df_static['type'] = 'Static'
df_dynamic['type'] = 'Dynamic'

combined_df = pd.concat([df_static, df_dynamic], ignore_index=True)
combined_df.to_csv("full_benchmark_combined.csv", index=False)
print("Combined benchmark data written to full_benchmark_combined.csv")
```

```python
# Method to save all comparison visuals as PNG files
def save_all_visuals():
    # A* Path Lengths
    plt.figure(figsize=(12,6))
    plt.plot(df_static['run'], df_static['aco_astar_length'], label='ACO (A*) Static', color='blue', linestyle='--')
    plt.plot(df_dynamic['run'], df_dynamic['aco_astar_length'], label='ACO (A*) Dynamic', color='blue')
    plt.plot(df_static['run'], df_static['ga_astar_length'], label='GA (A*) Static', color='red', linestyle='--')
    plt.plot(df_dynamic['run'], df_dynamic['ga_astar_length'], label='GA (A*) Dynamic', color='red')
    plt.plot(df_static['run'], df_static['hybrid_astar_length'], label='Hybrid (A*) Static', color='magenta', linestyle='--')
    plt.plot(df_dynamic['run'], df_dynamic['hybrid_astar_length'], label='Hybrid (A*) Dynamic', color='magenta')
    plt.xlabel('Run')
    plt.ylabel('Path Length (meters)')
    plt.title('A* Router: Path Lengths (Static vs Dynamic Hyperparameters)')
    plt.legend()
    plt.savefig('astar_pathlengths_comparison.png')
    plt.close()

    # A* Execution Times
    plt.figure(figsize=(12,6))
    plt.plot(df_static['run'], df_static['aco_astar_time'], label='ACO (A*) Static', color='blue', linestyle='--')
    plt.plot(df_dynamic['run'], df_dynamic['aco_astar_time'], label='ACO (A*) Dynamic', color='blue')
    plt.plot(df_static['run'], df_static['ga_astar_time'], label='GA (A*) Static', color='red', linestyle='--')
    plt.plot(df_dynamic['run'], df_dynamic['ga_astar_time'], label='GA (A*) Dynamic', color='red')
    plt.plot(df_static['run'], df_static['hybrid_astar_time'], label='Hybrid (A*) Static', color='magenta', linestyle='--')
    plt.plot(df_dynamic['run'], df_dynamic['hybrid_astar_time'], label='Hybrid (A*) Dynamic', color='magenta')
    plt.xlabel('Run')
    plt.ylabel('Execution Time (seconds)')
    plt.title('A* Router: Execution Times (Static vs Dynamic Hyperparameters)')
    plt.legend()
    plt.savefig('astar_executiontimes_comparison.png')
    plt.close()

    # Dijkstra Path Lengths
    plt.figure(figsize=(12,6))
    plt.plot(df_static['run'], df_static['aco_dijkstra_length'], label='ACO (Dijkstra) Static', color='blue', linestyle='--')
    plt.plot(df_dynamic['run'], df_dynamic['aco_dijkstra_length'], label='ACO (Dijkstra) Dynamic', color='blue')
    plt.plot(df_static['run'], df_static['ga_dijkstra_length'], label='GA (Dijkstra) Static', color='red', linestyle='--')
    plt.plot(df_dynamic['run'], df_dynamic['ga_dijkstra_length'], label='GA (Dijkstra) Dynamic', color='red')
    plt.plot(df_static['run'], df_static['hybrid_dijkstra_length'], label='Hybrid (Dijkstra) Static', color='magenta', linestyle='--')
    plt.plot(df_dynamic['run'], df_dynamic['hybrid_dijkstra_length'], label='Hybrid (Dijkstra) Dynamic', color='magenta')
    plt.xlabel('Run')
    plt.ylabel('Path Length (meters)')
    plt.title('Dijkstra Router: Path Lengths (Static vs Dynamic Hyperparameters)')
    plt.legend()
    plt.savefig('dijkstra_pathlengths_comparison.png')
    plt.close()

    # Dijkstra Execution Times
    plt.figure(figsize=(12,6))
    plt.plot(df_static['run'], df_static['aco_dijkstra_time'], label='ACO (Dijkstra) Static', color='blue', linestyle='--')
    plt.plot(df_dynamic['run'], df_dynamic['aco_dijkstra_time'], label='ACO (Dijkstra) Dynamic', color='blue')
    plt.plot(df_static['run'], df_static['ga_dijkstra_time'], label='GA (Dijkstra) Static', color='red', linestyle='--')
    plt.plot(df_dynamic['run'], df_dynamic['ga_dijkstra_time'], label='GA (Dijkstra) Dynamic', color='red')
    plt.plot(df_static['run'], df_static['hybrid_dijkstra_time'], label='Hybrid (Dijkstra) Static', color='magenta', linestyle='--')
    plt.plot(df_dynamic['run'], df_dynamic['hybrid_dijkstra_time'], label='Hybrid (Dijkstra) Dynamic', color='magenta')
    plt.xlabel('Run')
    plt.ylabel('Execution Time (seconds)')
    plt.title('Dijkstra Router: Execution Times (Static vs Dynamic Hyperparameters)')
    plt.legend()
    plt.savefig('dijkstra_executiontimes_comparison.png')
    plt.close()

    print("All comparison plots saved as PNG files in the current directory.")

# Call the method to save all visuals
save_all_visuals()
```
